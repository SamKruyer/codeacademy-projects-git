```python
#import CSV file, store as dictionary
import csv

with open('insurance.csv') as insurance_data:
    insuranceCSV = csv.DictReader(insurance_data)
    insurance_dict = {}
    key = 1
    for row in insuranceCSV:
        for sub_dict_key in row:
            row[sub_dict_key] = conv(row.get(sub_dict_key))
        insurance_dict[key] = row
        key +=1
```

```python
#takes dictionary where value pairs are dictionaries along with key, returns list of all responses for that key in the value dictionaries
def dict_to_list(dictionary, target_key):
    key_list = dictionary.keys()
    value_list = []
    for key in key_list:
        value = dictionary[key].get(target_key)
        value_list.append(value)
    return value_list

#convert strings that are numbers to type float -- called when making dictionary to convert strings from CSV import into floats for data analysis
def conv(s):
    try:
        s=float(s)
    except ValueError:
        pass    
    return s

#takes list of numbers and returns average rounded to 2 decimal places
def number_average(list):
    total = 0
    for element in list:
        total += element
    average = round(total/len(list), 2)
    return average

#takes list and returns distribution of responses in dictionary as percentages
def get_distribution(list):
    total_responses = len(list)
    response_dictionary = {}
    for item in list:
        if item in response_dictionary:
            response_dictionary[item] = response_dictionary.get(item) + 1
        else:
            response_dictionary[item] = 1
    for key in response_dictionary:
        response_dictionary[key] = round(response_dictionary.get(key)/total_responses*100, 2)

    #sorts by keys for ease of reading
    sorted_dict = dict(sorted(response_dictionary.items()))
    
    return sorted_dict
```

```python
#exploratory data analysis along the lines of children
def parents_no_kids_analysis(dictionary, variable):
    parents_list = []
    no_kids_list = []
    test_key = list(dictionary.keys())[0]
    for entry in dictionary:
        kids = dictionary[entry].get('children')
        if kids > 0:
            parents_list.append(dictionary[entry].get(variable))  
        else:
            no_kids_list.append(dictionary[entry].get(variable))

    if isinstance(dictionary[test_key].get(variable), float):
        parent_average = str(number_average(parents_list))
        no_kids_average = str(number_average(no_kids_list))
        print('The average ' + variable + ' of parents is ' + parent_average + ' compared to an average of ' + no_kids_average + ' for people without kids.')
    else:
        parent_dist = str(get_distribution(parents_list))
        no_kids_dist = str(get_distribution(no_kids_list))
        print('The distribution of parents by ' + variable + ' is: ' + parent_dist + '\nThe distribution of non-parents by ' + variable + ' is: ' + no_kids_dist)

#average age of parents vs. non-parents
parents_no_kids_analysis(insurance_dict, 'age')

#average bmi of parents vs. non-parents
parents_no_kids_analysis(insurance_dict, 'bmi')

#average cost of parents vs. non-parents
parents_no_kids_analysis(insurance_dict, 'charges')

#parents vs. non-parents by region
parents_no_kids_analysis(insurance_dict, 'region')

#parents vs. non-parents by smoking
parents_no_kids_analysis(insurance_dict, 'smoker')

#clearly there is no real difference between people with kids and without kids aside from slight differences in insurance cost!
```
